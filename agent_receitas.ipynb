# Agente de receitas
- Minha ideia é fazer um agent para receitas, colocarei receitas de algumas comidas nele pra ter sempre salvo

```python
# Adicionando imports

from langchain_community.document_loaders import DirectoryLoader, TextLoader

#dados

# procurar os arquivos .md 
loader = DirectoryLoader(
    "./data/receitas",
    glob="**/*.md",
    loader_cls=lambda path: TextLoader(path, encoding="utf-8")
    )

# carrega os arquivos .md achados
docs = loader.load()

# printa as receitas presentes:
for doc in docs:
    titulo = doc.page_content.split("\n")[0].replace("#", "").strip()
    print(f"--- Receita disponiveis: {titulo} ---")

```
